```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score

df = pd.read_csv('/content/simple_cyber_messages.csv')

display(df.head())

print('Dataset shape:', df.shape)
```

Manual Threat Identification

```python
manual_table = df.head(10)[
    ['message_id', 'message', 'correct_category']
].copy()

manual_table['important_words'] = [
    'meeting, scheduled',
    'suspended, click here, verify, password',
    'urgently, bank, verify',
    'executable, trojan, infect',
    'encrypted, bitcoin, restore files',
    'library, returned',
    'prize, login, claim',
    'spyware, attachment',
    'locked files, ransom, restore access',
    'official university portal'
]

display(manual_table)
```

Keyword Lists and Rule-Based Classifier

```python
RULE_KEYWORDS = {
    'Phishing': [
        'click here', 'verify', 'password', 'urgent',
        'otp', 'prize', 'bank', 'login', 'blocked'
    ],

    'Malware': [
        'malware', 'virus', 'infected', 'attachment',
        'trojan', 'spyware', 'executable'
    ],

    'Ransomware': [
        'encrypted', 'ransom', 'bitcoin', 'restore files',
        'locked files', 'unlock', 'payment demand'
    ]
}


def classify_message(text):
    text = str(text).lower()

    scores = {
        category: sum(word in text for word in words)
        for category, words in RULE_KEYWORDS.items()
    }

    predicted = max(scores, key=scores.get)

    if scores[predicted] == 0:
        return 'Safe', '', 0

    matched = [
        word for word in RULE_KEYWORDS[predicted]
        if word in text
    ]

    return predicted, ', '.join(matched), scores[predicted]


print(classify_message(
    'Urgent: click here to verify your password'
))
```

Complete Dataset Classification and Accuracy

```python
results = df['message'].apply(classify_message)

df[['predicted_category', 'matched_keywords', 'keyword_score']] = pd.DataFrame(
    results.tolist(),
    index=df.index
)

df['correct'] = df['correct_category'] == df['predicted_category']

accuracy = accuracy_score(
    df['correct_category'],
    df['predicted_category']
)

display(df[
    ['message_id',
     'correct_category',
     'predicted_category',
     'matched_keywords',
     'correct']
])

print('Correct predictions:', df['correct'].sum())
print('Incorrect predictions:', (~df['correct']).sum())
print(f'Accuracy: {accuracy * 100:.2f}%')
```

Risk Level, Recommended Action and Bar Chart

```python
# Create a function to assign risk level and recommended action.
def add_risk_and_action(row):

    # If the predicted category is Safe, assign low risk.
    if row['predicted_category'] == 'Safe':
        return 'Low', 'Continue normal caution.'

    # If the message is ransomware or has two or more risky keywords,
    # assign a high risk level.
    if (
        row['predicted_category'] == 'Ransomware'
        or row['keyword_score'] >= 2
    ):
        return (
            'High',
            'Do not click/open; isolate if required and report.'
        )

    # Assign medium risk to all other suspicious messages.
    return (
        'Medium',
        'Do not interact; verify the sender and report.'
    )


# Apply the function to each row of the DataFrame.
risk_output = df.apply(add_risk_and_action, axis=1)

# Add the risk levels and recommended actions as new DataFrame columns.
df[['risk_level', 'recommended_action']] = pd.DataFrame(
    risk_output.tolist(),
    index=df.index
)


# Create empty lists to store risk levels and recommended actions.
risk_levels = []
recommended_actions = []


# Read each row of the DataFrame one by one.
for index, row in df.iterrows():

    # Get the predicted category of the current message.
    category = row['predicted_category']

    # Get the keyword score of the current message.
    score = row['keyword_score']

    # If the message is safe, assign a low risk level.
    if category == 'Safe':
        risk = 'Low'
        action = 'Continue normal caution.'

    # If the message is ransomware or has two or more risky keywords,
    # assign a high risk level.
    elif category == 'Ransomware' or score >= 2:
        risk = 'High'
        action = 'Do not click/open; isolate if required and report.'

    # Assign medium risk to all other suspicious messages.
    else:
        risk = 'Medium'
        action = 'Do not interact; verify the sender and report.'

    # Store the risk level in the list.
    risk_levels.append(risk)

    # Store the recommended action in the list.
    recommended_actions.append(action)


# Add the risk levels as a new DataFrame column.
df['risk_level'] = risk_levels

# Add the recommended actions as a new DataFrame column.
df['recommended_action'] = recommended_actions


# Set the order in which categories should appear in the graph.
category_order = ['Safe', 'Phishing', 'Malware', 'Ransomware']


# Count the number of messages in each predicted category.
category_counts = df['predicted_category'].value_counts()


# Arrange the categories in the required order.
# If a category is missing, its count will be set to 0.
category_counts = category_counts.reindex(
    category_order,
    fill_value=0
)


# Create a bar graph of the predicted message categories.
category_counts.plot(
    kind='bar',
    color='skyblue',
    edgecolor='black'
)


# Add a title to the graph.
plt.title('Predicted Message Categories')

# Add a label to the horizontal axis.
plt.xlabel('Category')

# Add a label to the vertical axis.
plt.ylabel('Number of Messages')

# Keep the category labels horizontal.
plt.xticks(rotation=0)

# Adjust spacing so that all graph elements are visible.
plt.tight_layout()

# Display the graph.
plt.show()
```
